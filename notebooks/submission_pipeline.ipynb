# LIRA V44 Kaggle Submission Pipeline

This notebook assumes all V44 assets have already been prepared and packaged offline. It does not rebuild FAISS, BM25, case profiles, or query-view caches.

```python
from pathlib import Path
import lira_pipeline

cfg = lira_pipeline.V44Config()
lira_pipeline.validate_kaggle_manifest(require_bundle=True)
bundle = lira_pipeline.load_model_bundle()
resources = lira_pipeline.load_resources(cfg, require_reranker=True)
submission = lira_pipeline.predict_split(resources, bundle, split='test')
output_path = Path('/kaggle/working/submission.csv')
submission.to_csv(output_path, index=False)
print(output_path)
submission.head()
```
